### 🟠 Nível 3: Curiosidades e Estatísticas Simples

Vamos usar funções matemáticas básicas para entender a lista.

**Exercício 7:**
Qual é a **média das notas** (`Rating`) de todos os 250 filmes? E qual é a média dos anos de lançamento?

*Dica: Use o comando `.mean()` na coluna desejada.*

```python
# Seu código aqui
media_notas = df['Rating'].mean()

media_anos = df['Year'].mean()

print(f'A média das notas  de todos os 250 filmes é {media_notas:.2f}')
print(f'A média dos anos de Lançamento de todos os 250 filme3s é {media_anos:.0f}')
```

### 🔴 Nível 4: Um pouco de Visualização (Gráficos)

Uma imagem vale mais que mil palavras. Vamos criar gráficos simples.

**Exercício 9:**
Crie um **gráfico de barras** mostrando a contagem de filmes por ano (`Year`).
*Pergunta:* Teve algum ano que teve muitos filmes na lista?

*Dica: Use `df['Year'].value_counts().head(10).plot(kind='bar')`*

**Exercício 10:**

*Dica: Use `df['Rating'].plot(kind='hist', bins=20)`*
